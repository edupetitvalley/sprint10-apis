# Level 1: Basic exploration with a sandbox API

From a Jupyter Notebook, you will complete the following exercises using Python's `requests` library:


## 1. **Query the JSONPlaceholder public API** using the `GET` method to obtain:
   * A list of posts (`/posts`)
   * A list of users (`/users`)
   * A list of to-dos (`/todos`)


```python
import requests
import pandas as pd
import json

# BASE_URL = "https://jsonplaceholder.typicode.com"
BASE_URL = "https://jsonplaceholder.typicode.com"
# BASE_URL = "https://typicode.com"


posts_response = requests.get(f"{BASE_URL}/posts")
users_response = requests.get(f"{BASE_URL}/users")
todos_response = requests.get(f"{BASE_URL}/todos")
```


## 2. **Display on the screen:**
   * The total quantity of each resource.
   * The status code of each request.


```python
# The total quantity of each resource.
print("posts count: " , len(posts_response.json()))
print("users count: " , len(users_response.json()))
print("todos count: " , len(todos_response.json()))

# The status code of each request.
print("posts status code: " , posts_response.status_code)
print("users status code: " , users_response.status_code)
print("todos status code: " , todos_response.status_code)
```

## 3. **Make a request to a non-existent post** to trigger a `404` error and display the received status code.



```python
# Making the GET request to a negative post id
response = requests.get(f"{BASE_URL}/posts/10000000000000000000000")

# Status
print(f"Status Code: {response.status_code}")
```

## 4. **Make a POST request** to create a new fictional post. Include a title, body text, and a `userId`.
   * Display the JSON response.
   * Display the status code.



```python
# Display the JSON response.
response_post = requests.post(f"{BASE_URL}/posts", json={
    "title": "titulo inventado",
    "body": "body inventado.",
    "userId": 123
})

print("JSON response_post : ", response_post.json())
```

```python
# Display the status code. 
print("status code : ", response_post.status_code)
```

---
## 5. **Make a PATCH request** to partially modify an existing post.
   * Display the JSON response.
   * Display the status code.


```python
# we get the post id from previous post action
post_id = response_post.json()["id"]

# Display the JSON response PATCH. We include the post id in the url path
response_patch = requests.patch(f"{BASE_URL}/posts/{post_id}", json={
    "title": "titulo modificado",
    "body": "body modificado con patch.",
    "userId": 123
})

print("post_id: ", post_id)
print("JSON response_patch : ", response_patch.json())
```

```python
#    * Display the status code.
print("response_patch.status_code = ", response_patch.status_code)
```

---
## 6. **Make a DELETE request** on a post.
   * Display the JSON response.
   * Display the status code.

```python
# we get the post id from previous post action
post_id = response_post.json()["id"]

# Display the JSON response delete. We include the post id in the url path
response_delete = requests.delete(f"{BASE_URL}/posts/{post_id}")

print("post_id: ", post_id)
print("JSON response_delete : ", response_delete.json())
```

```python
#    * Display the status code.
print("response_delete.status_code = ", response_delete.status_code)
```

## 2. Read API's documentation

    2a. Review the available endpoints section In a markdown cell, note at least two different endpoints.

Available ClinicalTrials.gov API v2 Endpoints

- **`GET /api/v2/studies`**: Searches and returns a paginated list of clinical studies matching specific query parameters and filters (such as condition, intervention, location, or recruitment status).

- **`GET /api/v2/studies/{nctId}`**: Retrieves the complete study record for a single clinical trial using its unique 8-digit NCT identifier.

- **`GET /api/v2/studies/metadata`**: Returns the study data model, listing all available data elements and field definitions for a study record.


    2b. Check if it offers interesting filters or optional parameters. Note them down in markdown.


> Search & Query Parameters (query.*)
- query.cond. - Searches for specific medical conditions or diseases (e.g., "Cholangiocarcinoma" or "Ovarian Neoplasms").
- query.locn. Filters studies by geographic location text, such as city, state, or country (e.g., "Barcelona" or "Spain").
- query.spons. Searches by the lead sponsor or collaborating organization name (e.g., "AstraZeneca" or "Novartis").

> Strict Filter Parameters (filter.*). These parameters apply exact constraints to narrow your search results:
- filter.overallStatus. Filters by recruitment status using a comma-separated list of uppercase strings (e.g., "RECRUITING,COMPLETED"). Your uploaded sample contains statuses like "UNKNOWN" and "COMPLETED".
- filter.advanced. Applies complex boolean logic using Essie expression syntax (e.g., "AREA[Phase]PHASE3 AND AREA[EnrollmentCount]RANGE[100,MAX]").

> Output, Sorting & Pagination Parameters. Because the default payload contains dozens of nested modules (such as protocolSection, derivedSection, and resultsSection), these parameters help optimize your response:
- fields. Limits the returned JSON to a comma-separated list of specific fields, dramatically reducing payload size. You can use top-level names like "NCTId,BriefTitle,OverallStatus" or dot-notation paths like "protocolSection.identificationModule.nctId".
- sort. Orders the results using a comma-separated list of fields and directions (e.g., "LastUpdatePostDate:desc").


    2c. Verify that the response is in JSON format (it is the most suitable for this exercise).

      2d. Make a simple GET request: Display the response status code.


```python
print("response_nct.status_code = ", response_nct.status_code)
```

      Print some of the fields from the JSON response clearly.
> We will show the title and lead sponsor of 5 studies located in Barcelona


```python
params = {
    "query.locn": "Barcelona",
    "query.spons": "Novartis",  
    "fields": "BriefTitle,LeadSponsorName",
    "pageSize": 5,
}

headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}

response = requests.get(BASE_URL_NTC, params=params, headers=headers)
data = response.json()

#traverse the dictionary data object(studies) with get(dictionary, fallback)
for study in data.get("studies", []):
    protocol = study["protocolSection"]

    title = protocol["identificationModule"]["briefTitle"]
    lead_sponsor = protocol["sponsorCollaboratorsModule"]["leadSponsor"]["name"]

    print(f"Title:        {title}")
    print(f"Lead Sponsor: {lead_sponsor}")
```

## 4. **Convert the data into a pandas DataFrame:**
   * Display the first few rows of the DataFrame.

```python
# 1. Convert the nested list of studies into a DataFrame
df = pd.json_normalize(data["studies"])

# 2. Display the first few rows (Step 4)
df.head()
```
